# Sentinel 2

### Importing and loading Sentinel2 datasets

```python
# importing the sentinel2 datasets:
## creating filepaths
data_folder = "data"
S2_2022_fp = os.path.join(data_folder, "S2_composite_2022.tif") #creates a filepath to the S2 composite of 2022
S2_2024_fp = os.path.join(data_folder, "S2_composite_2024.tif") #creates a filepath to the S2 composite of 2024
```

```python
# load all bands of the sentinel2 datasets using rasterio
with rasterio.open(S2_2022_fp) as src:
    img_2022 = src.read()

with rasterio.open(S2_2024_fp) as src:
    img_2024 = src.read()
```

```python
# Check data from sentinel2 datasets
## check dataset of 2022
print(f"img_2022:")
print(f"   - Type: {type(img_2022)}")  #checks if dataset is a NumPy array
print(f"   - Data Type: {img_2022.dtype}")
print(f"   - Dimension: {img_2022.ndim}")
print(f"   - Shape: {img_2022.shape}") # gives numbers of (channels, height, width)

## check dataset of 2024
print(f"img_2024:")
print(f"   - Type: {type(img_2024)}")  #checks if dataset is a NumPy array
print(f"   - Data Type: {img_2024.dtype}")
print(f"   - Dimension: {img_2024.ndim}")
print(f"   - Shape: {img_2024.shape}")
```

### Background map

```python
# make background map
## define a normalization function
def normalize(array, vmin=0, vmax=3500):
    """
    normalizes the array by clipping it to the boundaries 0 and 3500 and scaling it to 0-1 range

    Parameters
    ----------
    array : float
        array of interest

    vmin : int
        lower boundary

    vmax : int
        upper boundary

    Returns
    -------
    float
        normalized array
    """
    clipped = np.clip(array, vmin, vmax)
    return (clipped-vmin)/(vmax-vmin)

## extract and normalize bands
Blue_2022 = normalize(img_2022[1]) # blue is band 2 but because the index starts at 0 it is 1
Green_2022 = normalize(img_2022[2]) # green = band 3
Red_2022 = normalize(img_2022[3]) # red = band 4

Blue_2024 = normalize(img_2024[1])
Green_2024 = normalize(img_2024[2])
Red_2024 = normalize(img_2024[3])

## stack the bands (tcc = true color composites)
tcc_2022 = np.stack((Red_2022, Green_2022, Blue_2022), axis =-1) #axis=-1 to set the channel as last, because np.stack expect Height, width, channels and not C,H,W
tcc_2024 = np.stack((Red_2024, Green_2024, Blue_2024), axis =-1)

## plot the background map 
fig, (ax1, ax2) = plt.subplots(1,2)
ax1.imshow(tcc_2022)
ax2.imshow(tcc_2024)

fig.suptitle("Backgroundmaps")
ax1.set_title("TCC_2022")
ax2.set_title("TCC_2024")
```

## NBR

### Cleaning and extracting bands

```python
# extract the bands 8 and 12 for each year
## for the calculations, later on, we'll only need Band 8 and 12 of each year
band_8_2022 = img_2022[7] # 7 bc it starts counting at 0 instead of 1
band_12_2022 = img_2022[11] # year at the end bc otherwise it's invalid
band_8_2024 = img_2024[7]
band_12_2024 = img_2024[11]
```

```python
# Lets check for NaN values in our bands (8 and 12 of each year)
print(f" 2022 band 8 has {np.isnan(band_8_2022).sum()} NaN values") #checks if there're any NaN values and if so how many
print(f" 2022 band 12 has {np.isnan(band_12_2022).sum()} NaN values")
print(f" 2024 band 8 has {np.isnan(band_8_2024).sum()} NaN values")
print(f" 2024 band 12 has {np.isnan(band_12_2024).sum()} NaN values")
```

```python
# change NaN values of the bands to 0

band_8_2022_noNaN = np.nan_to_num(band_8_2022)
band_12_2022_noNaN = np.nan_to_num(band_12_2022)
band_8_2024_noNaN = np.nan_to_num(band_8_2024)
band_12_2024_noNaN = np.nan_to_num(band_12_2024)
```

### Calculating NBR and Area without clipping

```python
# calculate NBR of both years
#NBR = (band 8 - band 12)/(band 8 + band 12)

## define the NBR calculation
def calculate_NBR(band_8, band_12):
    """
    calculates the normalized burn ratio (NBR)

    Parameters
    ----------
    band_8 : float
        Band of the Sentinel2 satellite that captures the NIR (near infrared) wavelengths
        
    band_12 : float
        Band of the Sentinel2 satellite that captures the SWIR2 (short waved infrared2) wavelengths

    Returns
    -------
    float
        Normalized burn ratio
    """
    numerator = band_8 - band_12
    denominator = band_8 + band_12
    NBR = np.zeros_like(numerator, dtype= float) # makes an output array that contains only zeros, that we can then fill with our NBR calculations
    NBR = np.divide(numerator, denominator, out = NBR, where = denominator!=0) #sets NBR as output, that means the calculations is only performed where the denominator isn't 0, otherwise it just stays 0
    return NBR

# calculate the NBR for both years
NBR_2022 = calculate_NBR(band_8_2022_noNaN, band_12_2022_noNaN)
NBR_2024 = calculate_NBR(band_8_2024_noNaN, band_12_2024_noNaN)
    
```

```python
# calculate the difference of the NBR of the two years
diff_NBR = NBR_2022 - NBR_2024
```

```python
# define a formula for calculating the area in km2 
def calculate_area(array):
    """
    calculates the area in km2

    Parameters
    ----------
    array : float
        array of the area of interest

    Returns
    -------
    float
        area of interest in square kilometers
    """
    pixel_resolution = 10
    pixel_area_m2 = pixel_resolution ** 2
    area_km2 = ((np.sum(array))*pixel_area_m2)/1000000
    return area_km2

# print the areas as an f-string with 2 decimals
print(f"The NBR 2022 area is {calculate_area(NBR_2022):.2f} km^2")
print(f"The NBR 2024 area is {calculate_area(NBR_2024):.2f} km^2")
print(f"The difference in area of the 2022 and 2024 NBRs is {calculate_area(diff_NBR):.2f} km^2")
```

### Plotting NBR without clipping

```python
# plot NBRs of both years
fig, (ax1, ax2) = plt.subplots(1,2)
img_NBR_2022 = ax1.imshow(NBR_2022)
img_NBR_2024 = ax2.imshow(NBR_2022)
ax1.set_title("NBR of 2022")
ax2.set_title("NBR of 2024")
fig.colorbar(img_NBR_2022, ax=ax1)
fig.colorbar(img_NBR_2024, ax=ax2)
fig.tight_layout()
ax1.axis("off")
ax2.axis("off")
```

```python
norm = tsn(vmin=np.nanpercentile(diff_NBR,2), vcenter = 0,  vmax=np.nanpercentile(diff_NBR, 98))
# vcenter = 0, sets the center of the colorbar at 0

img_diff_NBR = plt.imshow(diff_NBR, cmap="RdBu_r",norm = norm)
plt.colorbar(img_diff_NBR)
plt.title("Difference of NBRs between 2022 and 2024")
plt.axis("off")

# plt.savefig("Fotos_etc./NoClip_NBR_change", dpi =300, bbox_inches = "tight")
```

### clipping the NBR change and calculating the new area

```python
# clip NBR change to only positiv values
diff_NBR_clipped = np.clip(diff_NBR, 0, None) 
## 0 lower boundary and None for upper boundary
```

```python
# calculate the area of the clipped NBR change using our earlier defined fomrula for the area
print(f"The difference in area of the 2022 and 2024 NBRs is {calculate_area(diff_NBR_clipped):.2f} km^2")
```

### plot the clipped NBR change

```python
# ploting the clipped NBR change
img_diff_NBR_clipped = plt.imshow(diff_NBR_clipped, cmap="Reds",vmin=np.nanpercentile(diff_NBR_clipped,2), vmax=np.nanpercentile(diff_NBR_clipped, 98))
plt.colorbar(img_diff_NBR_clipped)
plt.title("Clipped change of NBRs between 2022 and 2024")
plt.axis("off")


# plt.savefig("Fotos_etc./NBR_change_2022_2024", dpi =300, bbox_inches = "tight")
```

#### categorize and plot the extent of the NBR change

low severity: 0.1<= NBR change <= 0.27

moderate severity: 0.27 < NBR change <= 0.66

high severity: 0.66 < NBR change

```python
# categorize NBR extent
severity = np.full_like(diff_NBR_clipped, np.nan, dtype= float)

severity[((diff_NBR_clipped>= 0.1) & (diff_NBR_clipped<= 0.27))] = 1 # categorie 1 is low severity
severity[((diff_NBR_clipped>0.27) & (diff_NBR_clipped<= 0.66))] = 2 # categorie 2 is moderate severity
severity[(diff_NBR_clipped> 0.66)] = 3 # categorie 3 is high severity
```

```python
# plot the categorized NBR change severities
cmap = lcm(["yellow", "orange", "red"]) # cmap defined here and not in plt.imshow bc it's needed to set the norm: cmap.N
bounds = [0.5,1.5,2.5,3.5] # defines numerical intervalls in which the classes are centered
norm = bnorm(bounds, cmap.N) #cmap.N gives numberys of colors, here 3 because cmap contains 3 colors
plt.imshow(tcc_2022, alpha =0.8) # alpha=0.8 makes it slightly transparent
img_severity = plt.imshow(severity, cmap = cmap, norm=norm )
cbar = plt.colorbar(img_severity, ticks = [1,2,3])
cbar.ax.set_yticklabels(["Low", "Moderate", "High"])
plt.title("Severities of NBR change")
plt.axis("off")

# plt.savefig("Fotos_etc./NBR_change_2022_2024_cat", dpi =300, bbox_inches = "tight")
```

## Strong change for Sentinel2 datasets

### Calculate cosine similarity and change

```python
# Handle NAN and scale it to the range -1 to 1
def scaled(array):
    """
    set NaN values to 0 and scales it to the range -1 to 1

    Parameters
    ----------
    array : float
        array of interest

    Returns
    -------
    float
        scaled array to range -1 to 1 and no NaNs in the array
    """
    formula = ((np.nan_to_num(array)/5000)-1)
    return formula

S2_scaled_2022 = scaled(img_2022)
S2_scaled_2024 = scaled(img_2024)
```

```python
# Calculate the Vector Lengths (Norm)
def norm(array):
    """
    calculates the vector Lengths (Norm)

    Parameters
    ----------
    array : float
        array of interest

    Returns
    -------
    float
        the Norm of the Array
    """
    norm = np.sqrt(np.sum(array**2, axis=0))
    return norm

S2_norm_2022 = norm(S2_scaled_2022)
S2_norm_2024 = norm(S2_scaled_2024)
```

```python
# normalize the unit length
S2_unit_2022 = S2_scaled_2022/S2_norm_2022
S2_unit_2024 = S2_scaled_2024/S2_norm_2024
```

```python
# calculate cosine similarity
## determines the similarity of datasets based on their direction
cosine_sim_S2 = np.sum(S2_unit_2022*S2_unit_2024, axis=0) # axis = 0 to sum it across all bands
```

```python
# calculate the change cosine
change_cosine_S2 = 1.0 - cosine_sim_S2
```

### Calculate and plot strong change

```python
# calculate the mean and standard deviation
mean_val_S2 = np.nanmean(change_cosine_S2)
std_val_S2 = np.nanstd(change_cosine_S2)

# calculate z-score
z_score_S2 = (change_cosine_S2 - mean_val_S2)/std_val_S2

# threshold and mask
threshold_S2 = np.percentile(z_score_S2,95) #threshold of 95 means anything over 95% is a strong change
strong_change_S2 = z_score_S2 > threshold_S2
```

```python
# ploting strong change next to NBR change
fig, (ax1, ax2) = plt.subplots(1,2)
fig.suptitle("Sentinel 2 datasets")
    
## strong change
ax1.imshow(tcc_2022, alpha = 0.8)
img_strong_change_S2 = ax1.imshow(np.where(strong_change_S2, 1, np.nan),cmap ="autumn")
# np.where(,1,np.nan) makes false values of the boolean mask invisible
ax1.set_title("strong change")
ax1.axis("off")

## NBR change
img_diff_NBR_2 = ax2.imshow(diff_NBR_clipped, cmap="Reds",vmin=np.nanpercentile(diff_NBR_clipped,2), vmax=np.nanpercentile(diff_NBR_clipped, 98) )
ax2.set_title("Change in NBR")
fig.colorbar(img_diff_NBR_2, ax=ax2)
fig.tight_layout()
ax2.axis("off")

# fig.savefig("Fotos_etc./S2_StrongChange_NBR",dpi=300, bbox_inches = "tight")
```
